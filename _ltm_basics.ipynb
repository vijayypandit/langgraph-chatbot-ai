Creating a Long term memory store using LangGraph.
 The store is used to store the data in a key-value format. 
 The data can be retrieved using the key. The store can be used to store any type of data.
 User can create multiple namespaces to store different types of data. 
 The store can be used to store data in a persistent way. The data can be retrieved even after the program is closed and reopened.

```python
from langgraph.store.memory import InMemoryStore
```

```python
#creating a store 
store = InMemoryStore()
```

```python
#create namespace 
namespace = ("user","u1")
```

Creatingh Memory 

```python
store.put(namespace,"1",{"data":"User likes travelling"})
store.put(namespace,"2",{"data":"User likes Pasta"})
```

```python
#Getting the data from the store 
store.get(namespace,"2")
```

```python
#Printing all the data from the store using search method
items = store.search(namespace)

for item in items:
    print(item.value)
```

```python
#Semantic Search

# Gemini Embedding Model Imports
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import FAISS
from dotenv import load_dotenv
import os

# Load environment variables
load_dotenv(override=True)

# Initialize Gemini Embeddings
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")
```

```python
store = InMemoryStore(index={'embed': embeddings, 'dims': 1536})
```

```python
namespace = ("users","u1")
```

```python
store.put(namespace,"1",{"data":"User likes travelling"})
store.put(namespace,"2",{"data":"User likes Pasta"})
store.put(namespace,"3",{"data":"User likes to play football"})
store.put(namespace,"4",{"data":"User likes to play cricket"})
store.put(namespace,"5",{"data":"User mostly likes to play basketball"})
```

```python
items = store.search(namespace,query = "which game user mostly likes to play ?",limit=1)

for item in items:
    print(item.value)
```
